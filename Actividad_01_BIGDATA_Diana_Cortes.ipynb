**EJERCICO 3: PRIMER ANALISIS DE VENTAS**

```python
import pandas as pd
import numpy as np

np. random. seed (42)
n = 200
ventas = pd. DataFrame ( {
"fecha": pd.date_range ("2026-01-01", periods=n, freq="D"),
"ciudad": np. random. choice ( ["Bogota", "Medellin", "Cali", "Barranquilla"], n) ,
"canal": np. random.choice ( ["Tienda", "Web", "App"], n),
"unidades": np. random. randint (1, 20, n) ,
"precio": np.random.choice ( [25000, 48000, 99000], n)})

ventas["total"] = ventas ["unidades"] * ventas ["precio"]
print (ventas. shape)
print (ventas . head (3) )
```

```python
# 1. Total vendido por canal (Tienda, Web, App)
total_canal = ventas.groupby("canal")["total"].sum()
print("--- 1. TOTAL VENDIDO POR CANAL ---")
print(total_canal)

# 2. Ticket promedio por ciudad
ticket_ciudad = ventas.groupby("ciudad")["total"].mean().sort_values(ascending=False)
print("\n--- 2. TICKET PROMEDIO POR CIUDAD ---")
print(ticket_ciudad.round(2))

# 3. Día de la semana en que se vende más
ventas["dia_semana"] = ventas["fecha"].dt.day_name()
ventas_dia = ventas.groupby("dia_semana")["total"].sum().sort_values(ascending=False)
print("\n--- 3. TOTAL VENDIDO POR DÍA DE LA SEMANA ---")
print(ventas_dia)
```

El canal **Tienda** lidera los ingresos totales ($44802000) y los **Viernes** concentran el mayor volumen de ventas ($18.737M), mientras que **Medellin** registra el ticket promedio más alto, por lo que se recomienda reforzar el inventario físico los fines de semana y replicar en Cali la estrategia comercial de Medellin.
