**EJERCICO 1: CLASIFICADOR DE NOTAS**

```python
# Crear la función
def clasificar_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota <= 4.4:
        return "Aprueba"
    else:
        return "Excelente"

# Lista de notas, recorrido con ciclo for y conteo de aprobados
notas = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0

for n in notas:
    estado = clasificar_nota(n)
    print(f"Nota: {n} -> {estado}")

    if estado == "Aprueba" or estado == "Excelente":
        aprobados += 1

print(f"\nTotal de estudiantes que aprobaron: {aprobados}")
```

**EJERCICIO 2: CARRITO DE COMPRAS**

```python
# 1. Crear el diccionario:(precio, cantidad)
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

# 2. Recorrer el diccionario y calcular el subtotal
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    subtotal += precio * cantidad

print(f"Subtotal: ${subtotal:,.0f}")

# 3. Aplicar 10% de descuento si el subtotal supera 200.000
if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0

subtotal_con_descuento = subtotal - descuento
print(f"Descuento (10%): -${descuento:,.0f}")

# 4. Sumar IVA del 19% e imprimir el total con formato de pesos
iva = subtotal_con_descuento * 0.19
total = subtotal_con_descuento + iva

print(f"IVA (19%): ${iva:,.0f}")
print(f"Total a pagar: ${total:,.0f}")
```

```python
import pandas as pd

df = pd. DataFrame ({
"empleado": ["Ana", "Luis", "Marta", "Juan", "Sofia"],
"area": ["Ventas", "TI", "Ventas", "TI", "RR. HH. "],
"salario": [2800000, 4500000, 3100000, 5200000, 3000000]})

df["salario_anual"] = df ["salario"] * 12
print (df[df["salario"] > 3000000])
print ("Promedio por area:", df. groupby ("area") ["salario"] .mean () .to_dict () )
```

```python
import numpy as np

# 1. Agregar 3 empleados nuevos usando pd.concat
nuevos_empleados = pd.DataFrame({
    "empleado": ["Carlos", "Elena", "Diego"],
    "area": ["Ventas", "RR. HH.", "TI"],
    "salario": [3400000, 3200000, 4900000]
})

df = pd.concat([df, nuevos_empleados])

# 2. Crear la columna 'bono': 10% del salario si el área es Ventas, 5% si no
df["bono"] = np.where(
    df["area"] == "Ventas",
    df["salario"] * 0.10,
    df["salario"] * 0.05
)

df["salario_anual"] = df["salario"] * 12
df["costo_mensual"] = df["salario"] + df["bono"]

print("--- TABLA DE EMPLEADOS COMPLETA ---")
print(df)

# 3. Calcular el costo total de nómina mensual por área
costo_nomina_area = df.groupby("area")["costo_mensual"].sum().sort_values(ascending=False)
print("\n--- COSTO TOTAL DE NÓMINA MENSUAL POR ÁREA ---")
print(costo_nomina_area)

# 4. ¿Quién es el empleado con mayor salario?
mayor_salario = df.loc[df["salario"].idxmax()]
print("\n--- EMPLEADO CON MAYOR SALARIO ---")
print(f"Empleado: {mayor_salario['empleado']} | Área: {mayor_salario['area']} | Salario: ${mayor_salario['salario']:,.0f}")
```
