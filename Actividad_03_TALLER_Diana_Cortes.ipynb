**TEMA 3_OPERADORES:Problemas: ¿Te alcanza para el café?**

```python
presupuesto = 200000
precio_cafe = 8500
cafes_semana = 4

# 1. ¿Cuánto gastas al mes? (4 semanas)
gasto_mes = precio_cafe * cafes_semana * 4

# 2. ¿Cuánto te sobra?
sobra = presupuesto - gasto_mes

# 3. ¿Te alcanza? (True o False)
alcanza = presupuesto >= gasto_mes

print(gasto_mes, sobra, alcanza)
```

**TEMA 6_CONDICIONALES:Problema: ¿Te toca envio GRATIS?**

```python
total = 1500
es_miembro = False
compras = 2

# 1. Envío: más de 1000 -> gratis; miembro -> gratis; si no -> $99
if total > 1000:
    print("Envío gratis (compra mayor a $1000)")
elif es_miembro:
    print("Envío gratis (beneficio miembro)")
else:
    print("Costo de envío: $99")

# 2. Nivel: 10 o más -> Oro; de 5 a 9 -> Plata; menos -> Bronce
if compras >= 10:
    print("Nivel: Oro")
elif compras >= 5:
    print("Nivel: Plata")
else:
    print("Nivel: Bronce")
```
